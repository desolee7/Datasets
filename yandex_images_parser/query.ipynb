Major Crack (Большая трещина): +
сквозная трещина на стене здания, широкая трещина на фасаде, трещина в кирпичной стене крупным планом

Peeling (Отслаивание краски): +
вздутие краски на фасаде
облупившаяся краска на стене, шелушение фасадной краски

Spalling (Разрушение поверхности): 4Х50х10
скол бетона на фасаде +
разрушение железобетона +
разрушение кирпичной кладки на фасаде здания +
разрушение кирпичной кладки наружных стен +


Minor Crack (Мелкая трещина):
волосяная трещина на штукатурке, паутина трещин на бетоне, усадочная трещина на фасаде

Algae (Водоросли): зеленый налет на стене дома, плесень на фасаде здания, биопоражение кирпичной кладки

Stain (Пятна): ржавое пятно на стене, потек на фасаде от воды, темное пятно на бетонной стене, сажа на фасаде здания

Corrosion (Коррозия): ржавчина на металлическом фасаде, коррозия сайдинга, ржавые крепления на стене

Высолы (Efflorescence): белый налет на кирпиче, высолы на фасаде здания, солевые разводы на бетоне
