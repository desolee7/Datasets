```python
import os

def get_image_count(folder_path: str, recursive: bool = True) -> int:
    """
    Возвращает количество изображений в папке
    
    Args:
        folder_path: путь к папке
        recursive: учитывать ли подпапки
    
    Returns:
        количество изображений
    """
    extensions = ('.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp')
    count = 0
    
    if not os.path.exists(folder_path):
        print(f"Ошибка: Папка не существует: {folder_path}")
        return 0
    
    if recursive:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                if file.lower().endswith(extensions):
                    count += 1
    else:
        for file in os.listdir(folder_path):
            file_path = os.path.join(folder_path, file)
            if os.path.isfile(file_path) and file.lower().endswith(extensions):
                count += 1
    
    return count


FOLDER_PATH = "C:/datasets/hesh"

total_images = get_image_count(FOLDER_PATH, recursive=True)
print(f"Всего изображений в папке: {total_images}")
```
