```python
import os
import shutil
from tqdm import tqdm

def collect_all_images(root_folder):
    extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.tif', '.webp',
                  '.JPG', '.JPEG', '.PNG', '.BMP', '.TIFF', '.TIF', '.WEBP'}
    image_paths = []
    for root, dirs, files in os.walk(root_folder):
        for file in files:
            if os.path.splitext(file)[1] in extensions:
                image_paths.append(os.path.join(root, file))
    return image_paths

def rename_images_in_copy(source_folder, dest_folder):
    source_folder = os.path.abspath(source_folder)
    dest_folder = os.path.abspath(dest_folder)
    
    if os.path.exists(dest_folder):
        print(f"Удаление: {dest_folder}")
        shutil.rmtree(dest_folder)
    
    print(f"Создание копии: {dest_folder}")
    image_paths = collect_all_images(source_folder)
    print(f"Найдено изображений: {len(image_paths)}")
    
    print("Копирование и переименование...")
    
    success = 0
    for counter, src_path in enumerate(tqdm(image_paths), start=1):
        try:
            ext = os.path.splitext(src_path)[1]
            new_name = f"{counter:06d}{ext}"
            rel_path = os.path.relpath(src_path, source_folder)
            rel_dir = os.path.dirname(rel_path)
            dest_dir = os.path.join(dest_folder, rel_dir)
            dest_path = os.path.join(dest_dir, new_name)
            
            os.makedirs(dest_dir, exist_ok=True)
            shutil.copy2(src_path, dest_path)
            success += 1
        except Exception as e:
            continue
    
    print(f"\nГотово! Скопировано: {success}, пропущено: {len(image_paths) - success}")
    print(f"Папка: {dest_folder}")

source_folder = r"clean_images"
dest_folder = r"C:/datasets/renamed"

rename_images_in_copy(source_folder, dest_folder)
```
